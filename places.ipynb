```python
import pandas as pd
import geopandas as gpd

# Read CSV
df = pd.read_csv("penn_club_timeline.csv")

# Convert to GeoDataFrame
gdf = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df["lon"], df["lat"]),
    crs="EPSG:4326"
)

# View GeoDataFrame
gdf.head()
```

```python
gdf.to_file('places.geojson')
```

```python
gdf.plot()
```
